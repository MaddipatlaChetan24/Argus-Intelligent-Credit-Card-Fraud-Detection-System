```python
from fraud_detection.predict import predict_transaction
```

```python
import pandas as pd

df = pd.read_csv("../data/raw/creditcard.csv")

df = df.drop_duplicates()
```

```python
X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
transaction = X.iloc[0].values
actual_class = y.iloc[0]

print("Actual class:", actual_class)
```

```python
result, probability = predict_transaction(transaction)

print("Prediction:", result)
print("Fraud probability:", probability)
```

```python
fraud_index = y[y == 1].index[0]

fraud_transaction = X.loc[fraud_index].values

actual_class = y.loc[fraud_index]

result, probability = predict_transaction(
    fraud_transaction
)

print("Actual class:", actual_class)
print("Prediction:", result)
print("Fraud probability:", probability)
```
